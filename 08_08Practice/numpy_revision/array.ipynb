### Array generation funtion

```
np.random.rand(10)
# the below one is a vector , simple rule if we have one bracket then Vector , if two then Matrix
# random function give values between 0 & 1 ,this is called normalisation
```

```
np.random.randn(10)
# for values between 3 & -3 ,this is called standardisation
```

```
np.random.randint(6)
# gives one random no between 0 & 6
```

```
np.random.randint(9)
# returns random no between 0 & 9
```

```
np.random.randint(10,20)
```

```
np.random.randint(10,20,10)
# generates 10 random no. between 10 ,20 and values can be repeating
```
