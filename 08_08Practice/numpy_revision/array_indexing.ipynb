### BOOLEAN INDEXING

```
arr = np.arange(11,21)
arr
```

```
bool_index = arr % 2 ==0
bool_index
```

```
# dont execute this line more than once , as the arr data are already shorten , you have start from start of boolean indexing
arr = arr[bool_index]
arr
```
