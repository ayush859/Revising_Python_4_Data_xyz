### ARRAY OPERATIONS


#### Arithmetic operations

```
a1 = np.arange(1,6)
a2 = np.arange(6,11)
```

```
a1+a2
# works ,given both the arrays size are same
```

```
a1-a2
```

```
a1*a2
```

```
a1 / a2
```

```
a1 //a2
```

```
a1 ** a2
```

#### Broadcasting

```
l = [10,20,30,40,50]
arr = np.array((l))
```

```
arr + 10
```

```
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```
arr2+10
```

```
# there is no difference between broadcasting and arithmetic operations
```

### Advance Array Manipulation

#### Stacking Array

```
# In stacking we have to use VECTORS only , we can't use MATRIX
```

```
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```
np.vstack((a,b))
```

```
np.hstack((a,b))
# It basically joins two array
```

```
np.column_stack((a,b))
```

#### Splitting Array

```
c = np.arange(16).reshape(4,4)
c
```

```
np.hsplit(c,2)
# while splitting we have to take care that , we split in equal parts
```

```
np.hsplit(c,4)
```

```
a= np.vsplit(c,2)
```

```
a= np.vsplit(c,4)
a
```
