### Creating a DataFrame

```
''' We can create a datframe via Series , Dictionary '''
```

```
data = {
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, 30, 28, 35],
    "city": ["Bangalore", "Mumbai", "Delhi", "Chennai"],
    "salary": [50000, 60000, 55000, 70000]
}
df = pd.DataFrame(data)
```

```
data_list = [
    ["Alice", 25, "Bangalore", 50000],
    ["Bob", 30, "Mumbai", 60000],
    ["Charlie", 28, "Delhi", 55000],
    ["David", 35, "Chennai", 70000]
]
df2 = pd.DataFrame(data_list)
colums = ['name','age','city','salary']
df2= pd.DataFrame(data_list,columns = colums)
```

```
df2
```

```
df2['name']
```

```
'''If I want to target more than two columns then we have to create list inside list'''
```

```
df2[['name','city']]
```

### Creating a new Column

```
''' To add a new columns inside a Data Frame but while assigning value  take care we fill the required no. '''
df2['role']= ['Intern', 'Associate', 'Developer' , 'Associate']
```

```
df2
```

### Removing Columns

```
df2.drop('role',axis = 1)
'''This will not remove the data permanently or we have to use someplace to store this removed data and reassign it to df2 '''
```

```
df2
```

```
df2.drop('role',axis=1,inplace = True)
''' This will remove the data permanently  '''
```

```
df2
```

```
df2.drop(['city','salary'],axis=1)
```

```
''' If we want to remove the row ,then we have to make axis =0 '''
df2.drop(0,axis=0)
```

### Selectings Rows

### Selecting Subsets of Rows and Columns

```
df2.loc[[0,1]][['city','salary']]
```

### Conditional Selection

```
df2
```

```
''' Show people whose age is above 30 '''

df2[df2['age']>30]
```

```
df2[df2.iloc[:,1]>30]
```

```
''' Show people whose age is above 30 and city is Chennai '''
df2[(df2.iloc[:,1]>30 )& (df2.iloc[:,2]=='Chennai')]
```

```
df2[(df2['age']>30 )& (df2['city'] == 'Chennai')]
```
