```
import numpy as np
import pandas as pd
```

### GroupBy

```
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```
df
```

```
''' when we use group by , it create object not a dataframe which can't be viewed '''

cat = df.groupby('Category')
for i,v in cat:
    print(i)
    print(v)
```

```
''' Groupby Category and calculate the sum of Sales '''

cat = df.groupby('Category')['Sales'].sum()
cat
```

```
''' Groupby Store and calculate the sum of Sales '''

cat = df.groupby('Store')['Sales'].sum()
cat
```

##### When we GroupBy , we use it on categorical data  and not on continuous data

```
''' Groupby on multiple Columns '''
''' Group by Category and Store  and calculate sum of sales'''

cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```

### Aggregation

```
''' We have MEAN  MEDIAN  MODE MIN MAX COUNT STD'''
```

```
df['Sales'].mean()
```

```
df['Sales'].median()
```

```
''' we also have a Aggregation function  and within it we can pass any of the above except mode '''

df['Sales'].agg(['mean'])
```

```
df['Sales'].agg(['mean','median' ,'std','sum','min','max','count'])
```
