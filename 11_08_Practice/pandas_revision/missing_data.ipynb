```
import numpy as np
import pandas as pd
```

#### Finding Missing Data

```
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```
''' to check how many Null Values are there'''
df.isna()
```

```
'''For each label/column , we check total count of Null Value'''
df.isna().sum()
```

```
'to check for each columns do we any missing value  '
df.isna().any()
```

#### Removing Missing Data

```
''' Removing Data works on Rows basis '''
df
```

```
''' To remove rows that contain at least one NULL/NaN value (it will not remove the row which don't have any NULL Values') '''
df.dropna()
```

```
df
```

```
''' thresh specifies the minimum number of non-null values required in each row .Rows with at least the specified number of non-null values are retained. '''
df.dropna(thresh = 3)
```

```
df.dropna(thresh = 4)
```

```
df.dropna(thresh = 5)
```

#### Filling the missing Data

```
df
```

```
''' It makes sure that all the Null value are filled by X value.'''
df.fillna(0)
```

```
''' until and unless we write ( inplace = True ) ,No Changes will be made '''
df.fillna(11)
```

```
df
```

```
''' if for each columns we want to fill Null places by given value , '''

random_val = {'A':0,'B':120,'C':820,'D':970}
df.fillna(value= random_val)
```

##### dropna() => works ROW wise
##### fillna() => works COLUMNS wise

```
df.fillna(df.mean())
```
