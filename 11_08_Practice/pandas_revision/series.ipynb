## Pandas Series

### Creating Series

```
import numpy as np
import pandas as pd
```

```
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array(my_list)
dict = {1:10,2:20,3:30}
```

```
pd.Series(my_list)
```

```
pd.Series(my_list,labels)
```

```
pd.Series(arr,labels)
```

```
''' Series is 1D in itself '''
```

```
pd.Series(dict)
```

```
''' Multiple Series when come together , they act as DataFrame '''
```
