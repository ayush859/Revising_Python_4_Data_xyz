### Merging 2 dataframes

```
# DataFrame 1: Employee Information
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary Information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```
employees
```

```
salaries
```

```
''' we can merge the data on certain columns , but we have to make sure that column name is common in both table '''

pd.merge(employees,salaries,on='employee_id')
```

```
''' Merge can be done using inner / outer / left / right . '''
```

```
''' given how='inner' -> keeps only matching rows. '''

pd.merge(employees,salaries,on ='employee_id',how = 'inner')
```

```
''' given how='outer' -> keeps all rows from both DataFrames and for Missing Values we get NAN. '''
pd.merge(employees,salaries,on ='employee_id',how = 'outer')
```

```
'''how='left'  -> keeps all rows from the left DataFrame. and for other columns value are made NAN'''
pd.merge(employees,salaries,on ='employee_id',how = 'left')
```

```
'''how='left'  -> keeps all rows from the right DataFrame. and for other columns value are made NAN'''
pd.merge(employees,salaries,on ='employee_id',how = 'right')
```

##### We usually use INNER merge when we only need matching records, also it excludes unmatched rows and avoids NULLs caused by missing matches.

### Concatenation of 2 dataframes

```
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```
df1
```

```
df2
```

```
''' if we dont define the axis then concatenation is done via columns'''
pd.concat([df1,df2])
```

```
pd.concat([df1,df2],axis = 1)
```

#####m concat  → combine DataFrames
##### merge   → combine DataFrames based on a common column/key

### Joining of 2 dataframes

```
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```
df1
```

```
df2
```

```
df1.join(df2,how='outer')
```

```
df2.join(df1,how='outer')
```
